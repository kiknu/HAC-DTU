```python
from hac.eval import calculate_voxel_measure
from hac.data import get_model_folder

import numpy as np
import trimesh  # Commonly used for point-in-mesh checks

%load_ext autoreload
%autoreload 2
```

```python
model_dir = get_model_folder(3)
stl_file = model_dir / "asteroid3.stl"

true_mesh = trimesh.load_mesh(stl_file)
# Add some noise to the mesh for testing
rand_mesh = true_mesh.copy()
rand_mesh.vertices += np.random.normal(scale=0.01, size=rand_mesh.vertices.shape)

# Print shape and bounds of the meshes
print(f"True Mesh: {true_mesh.vertices.shape}, Bounds: {true_mesh.bounds}")
print(f"Rand Mesh: {rand_mesh.vertices.shape}, Bounds: {rand_mesh.bounds}")
```

```python
# Check is the voxel measure calculation works
# Compute the voxel measure score with true_mesh and itself (should be 1.0)
R = 1.0  # Example bounding cylinder radius
score_self = calculate_voxel_measure(true_mesh, true_mesh)
print(f"Voxel Measure Score (True Mesh vs True Mesh): {score_self}")
```

```python
# Tesing the voxel measure calculation
R = 1.0  # Example bounding cylinder radius
score = calculate_voxel_measure(true_mesh, rand_mesh)
print(f"Voxel Measure Score: {score:.4f}")
```
